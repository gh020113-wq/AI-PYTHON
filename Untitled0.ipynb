## Pandas `describe()` 함수를 이용한 요약 통계

`DataFrame.describe()` 함수는 숫자형(numerical) 데이터 열에 대한 요약 통계(예: 평균, 표준 편차, 최솟값, 최댓값, 사분위수 등)를 계산하여 보여줍니다. 범주형(categorical) 데이터에는 기본적으로 적용되지 않지만, `include='all'` 옵션을 사용하면 범주형 데이터에 대한 통계도 함께 볼 수 있습니다.

```python
# 숫자형 데이터에 대한 기본 요약 통계
print("\n숫자형 데이터 요약 통계:")
display(df.describe())

# 모든 데이터(숫자형, 범주형 포함)에 대한 요약 통계
print("\n모든 데이터 요약 통계 (include='all'):")
display(df.describe(include='all'))
```
